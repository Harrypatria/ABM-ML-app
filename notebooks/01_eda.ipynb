# Phase 2 — Exploratory Data Analysis (EDA)
## IBM HR Analytics Employee Attrition Dataset
**Objective:** Uncover patterns, distributions, correlations, and flight risk indicators.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-darkgrid')
os.makedirs('../reports/eda', exist_ok=True)

df = pd.read_csv('../data/raw/HR-Employee-Attrition.csv')
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# 1. Target Attrition Distribution
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
sns.countplot(data=df, x='Attrition', palette=['#06b6d4', '#f43f5e'], ax=ax[0])
ax[0].set_title('Attrition Frequency Count')

df['Attrition'].value_counts().plot.pie(autopct='%1.1f%%', colors=['#06b6d4', '#f43f5e'], ax=ax[1], explode=(0, 0.1))
ax[1].set_title('Class Imbalance (16.1% Positive)')
plt.savefig('../reports/eda/01_attrition_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# 2. Correlation Heatmap (Numeric features)
numeric_cols = df.select_dtypes(include=[np.number]).columns
plt.figure(figsize=(16, 12))
sns.heatmap(df[numeric_cols].corr(), cmap='coolwarm', annot=False, vmin=-1, vmax=1)
plt.title('Correlation Matrix of Numerical HR Attributes', fontsize=14)
plt.savefig('../reports/eda/02_correlation_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# 3. Box Plots: MonthlyIncome & Age by Attrition
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome', palette=['#06b6d4', '#f43f5e'], ax=ax[0])
ax[0].set_title('Monthly Income by Attrition Status')

sns.boxplot(data=df, x='Attrition', y='Age', palette=['#06b6d4', '#f43f5e'], ax=ax[1])
ax[1].set_title('Age Distribution by Attrition Status')
plt.savefig('../reports/eda/03_income_age_boxplots.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# 4. Stacked Bar Plots: Attrition Rate by OverTime & JobRole
fig, ax = plt.subplots(1, 2, figsize=(16, 6))
pd.crosstab(df['OverTime'], df['Attrition'], normalize='index').plot(kind='bar', stacked=True, color=['#06b6d4', '#f43f5e'], ax=ax[0])
ax[0].set_title('Attrition Ratio by Overtime Status')
ax[0].set_ylabel('Proportion')

pd.crosstab(df['JobRole'], df['Attrition'], normalize='index').plot(kind='barh', stacked=True, color=['#06b6d4', '#f43f5e'], ax=ax[1])
ax[1].set_title('Attrition Ratio by Job Role')
plt.savefig('../reports/eda/04_overtime_jobrole_attrition.png', dpi=300, bbox_inches='tight')
plt.show()
```
